```python
#question 5:
n=int(input("enter number of students: "))
for i in range(n):

    sname= input("enter name: ")
    sclass= input("enter class: ")

    m1= float(input("enter marks of sub1: "))
    m2= float(input("enter marks of sub2: "))
    m3= float(input("enter marks of sub3: "))
    m4= float(input("enter marks of sub4: "))
    m5= float(input("enter marks of sub5: "))

    total= m1+m2+m3+m4+m5
    percentage= (total/500)*100

    print("Name:", sname)
    print("Class:", sclass)
    print("Total Marks:", total)
    print("Percentage:", percentage, "%")

    if percentage >= 60:
            grade = 'A'
    elif percentage >= 50:
        grade = 'B'
    elif percentage >= 40:
        grade = 'C'
    elif percentage >= 33:
        grade = 'D'
    else:
        grade = 'Fail'

    print("grade:", grade)
```

```python
#question 4:
a = 10
b = 5

print("Initial Values")
print("a =", a)
print("b =", b)


a += b
print("\nAfter a += b, a= ",a)
print("a =", a)

a -= b
print("\nAfter a -= b, a= ",a)

a *= b
print("\nAfter a *= b, a= ",a)

a /= b
print("\nAfter a /= b, a= ",a)

a %= b
print("\nAfter a %= b, a= ",a)

a **= 2
print("\nAfter a **= b, a= ",a)

a //= 5
print("\nAfter a //= b, a= ",a)
```
