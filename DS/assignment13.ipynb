```python
#question 1:
import numpy as np

# Original array
arr = np.array([[6, -8, 73, -110],
                [np.nan, -8, 0, 94]])

# Replace NaN with 0
arr = np.nan_to_num(arr, nan=0)

# Interchange (swap) the two rows
arr[[0, 1]] = arr[[1, 0]]

# Interchange (swap) the 1st and 3rd columns
arr[:, [0, 2]] = arr[:, [2, 0]]

print(arr)
```

```python
#question 2:
import numpy as np

# Create a 3D array
arr = np.arange(24).reshape(2, 3, 4)

print("Original shape:", arr.shape)

# Move axis 0 to position 2
new_arr = np.moveaxis(arr, 0, 2)

print("New shape:", new_arr.shape)
print(new_arr)
```

```python
#question 3:
import numpy as np

# Sample array with NaN values
arr = np.array([[1, 2, np.nan],
                [4, np.nan, 6],
                [7, 8, 9]])

# Calculate column-wise averages ignoring NaN
col_avg = np.nanmean(arr, axis=0)

# Find indices of NaN values
inds = np.where(np.isnan(arr))

# Replace NaN values with corresponding column averages
arr[inds] = np.take(col_avg, inds[1])

print(arr)
```

```python
#question 5:
import numpy as np
from scipy import stats

# Create two NumPy 2D arrays
arr1 = np.array([[3, 4],
                 [5, 6]])

arr2 = np.array([[1, 0],
                 [7, 8]])

# Calculate average of the two arrays
avg = (arr1 + arr2) / 2

print("Average of NumPy arrays:\n", avg)

# Mean
mean_val = np.mean(avg)
print("Mean =", mean_val)

# Median
median_val = np.median(avg)
print("Median =", median_val)

# Mode
mode_val = stats.mode(avg, axis=None, keepdims=False)
print("Mode =", mode_val.mode)
```

```python
#question 6:
import numpy as np

#Using NumPy linalg.solve():
A = np.array([[1, -2, 3],
              [-1, 3, -1],
              [2, -5, 5]])

B = np.array([9, -6, 17])

X = np.linalg.solve(A, B)

print("Solution:")
print("x =", X[0])
print("y =", X[1])
print("z =", X[2])

#using inverse matrix:

A = np.array([[1, -2, 3],
              [-1, 3, -1],
              [2, -5, 5]])

B = np.array([[9],
              [-6],
              [17]])

A_inv = np.linalg.inv(A)

X = np.dot(A_inv, B)

print("Solution:")
print("x =", X[0][0])
print("y =", X[1][0])
print("z =", X[2][0])
```
