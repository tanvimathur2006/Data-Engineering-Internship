```python
#question 1:
import csv

# Create and open CSV file
with open("address_book.csv", "w", newline="") as file:
    writer = csv.writer(file)

    # Write column headings
    writer.writerow(["Name", "Address", "Mobile", "Email"])

    # Input 3 dummy records from user
    for i in range(3):
        print(f"\nEnter details for person {i+1}")

        name = input("Enter Name: ")
        address = input("Enter Address: ")
        mobile = input("Enter Mobile Number: ")
        email = input("Enter Email: ")

        # Write data into CSV file
        writer.writerow([name, address, mobile, email])

print("\nAddress book CSV file created successfully.")
```

```python
#question 2:
import sqlite3

# Connect to database
conn = sqlite3.connect("college.db")

# Create cursor object
cur = conn.cursor()

# 1) Create Tables

cur.execute("""
CREATE TABLE IF NOT EXISTS Student(
    StudentID INTEGER PRIMARY KEY,
    Name TEXT,
    Branch TEXT,
    Age INTEGER
)
""")

cur.execute("""
CREATE TABLE IF NOT EXISTS Course(
    CourseID INTEGER PRIMARY KEY,
    CourseName TEXT,
    Duration TEXT
)
""")

cur.execute("""
CREATE TABLE IF NOT EXISTS Marks(
    StudentID INTEGER,
    CourseID INTEGER,
    Marks INTEGER
)
""")


# 2) Insert Records

cur.execute("INSERT INTO Student VALUES (1, 'Rahul', 'CSE', 20)")
cur.execute("INSERT INTO Student VALUES (2, 'Neha', 'IT', 21)")
cur.execute("INSERT INTO Student VALUES (3, 'Aman', 'ECE', 19)")

cur.execute("INSERT INTO Course VALUES (101, 'Python', '3 Months')")
cur.execute("INSERT INTO Course VALUES (102, 'DBMS', '4 Months')")
cur.execute("INSERT INTO Course VALUES (103, 'Java', '5 Months')")

cur.execute("INSERT INTO Marks VALUES (1, 101, 85)")
cur.execute("INSERT INTO Marks VALUES (2, 102, 90)")
cur.execute("INSERT INTO Marks VALUES (3, 103, 78)")


# Save changes
conn.commit()


# 3) Different SELECT Operations

print("\nAll Students")
for row in cur.execute("SELECT * FROM Student"):
    print(row)

print("\nStudent Names")
for row in cur.execute("SELECT Name FROM Student"):
    print(row)

print("\nStudents Age Greater Than 19")
for row in cur.execute("SELECT * FROM Student WHERE Age > 19"):
    print(row)

print("\nMarks Greater Than 80")
for row in cur.execute("SELECT * FROM Marks WHERE Marks > 80"):
    print(row)

print("\nStudents Sorted By Age")
for row in cur.execute("SELECT * FROM Student ORDER BY Age"):
    print(row)

# 4) Update Some Data

cur.execute("UPDATE Student SET Age = 22 WHERE StudentID = 2")
cur.execute("UPDATE Marks SET Marks = 95 WHERE StudentID = 1")

conn.commit()


# 5) Delete Some Data

cur.execute("DELETE FROM Student WHERE StudentID = 3")
cur.execute("DELETE FROM Marks WHERE Marks < 80")

conn.commit()


# Final Data

print("\nFinal Student Table")
for row in cur.execute("SELECT * FROM Student"):
    print(row)

print("\nFinal Marks Table")
for row in cur.execute("SELECT * FROM Marks"):
    print(row)


# Close connection
conn.close()
```
